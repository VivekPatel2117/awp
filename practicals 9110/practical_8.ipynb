```python
import pandas as pd
import numpy as np
from apyori import apriori

```

```python

path = "/content/Market_Basket_Optimisation - Market_Basket_Optimisation (1).csv"
df= pd.read_csv(path)
```

```python
df.head()
```

```python
df.fillna(0,inplace=True)
```

```python
df.head()
```

```python
transactions = []

for i in range(len(df)):
    transactions.append([str(x) for x in df.iloc[i] if str(x) != ''])
```

```python
rules = apriori(transactions,
                min_support=0.003,
                min_confidence=0.2,
                min_lift=3,
                min_length=2)

results = list(rules)
```

```python
results
```

```python
rows = []

for item in results:
    pair = item.items
    support = item.support

    for stat in item.ordered_statistics:
        lhs = ', '.join(stat.items_base)
        rhs = ', '.join(stat.items_add)
        confidence = stat.confidence
        lift = stat.lift

        rows.append([lhs, rhs, support, confidence, lift])

df_final = pd.DataFrame(rows, columns=['lhs','rhs','support','confidence','lift'])
df_final.head()
```

```python
df_final.sort_values('lift', ascending=False).head(10)
```

```python
transactions = []

for i in range(0,len(df)):
    transactions.append([str(df.values[i,j]) for j in range(0,20) if str(df.values[i,j])!='0'])

transactions[0]
transactions[1]
```

```python
rules = apriori(transactions, min_support=0.003, min_confidence=0.2, min_lift=3, min_length=2)

Results = list(rules)
Results
```

```python
df_results = pd.DataFrame(Results)
df_results.head()

support = df_results.support
```

```python
first_values = []
second_values = []
third_values = []
fourth_value = []

for i in range(df_results.shape[0]):
    single_list = df_results['ordered_statistics'][i][0]
    first_values.append(list(single_list[0]))
    second_values.append(list(single_list[1]))
    third_values.append(single_list[2])
    fourth_value.append(single_list[3])
```

```python
lhs = pd.DataFrame(first_values)
rhs = pd.DataFrame(second_values)
confidence=pd.DataFrame(third_values,columns=['confidence'])
lift=pd.DataFrame(fourth_value,columns=['lift'])
```

```python
df_final = pd.concat([lhs,rhs,support,confidence,lift], axis=1)

df_final.fillna(value=' ', inplace=True)
df_final.head()
```

```python
df_final['lhs'] = df_final['lhs_item_1'] + str(', ') + df_final['lhs_item_2'] + str(', ') + df_final['lhs_item_3']
df_final['rhs'] = df_final['rhs_item_1'] + str(', ') + df_final['rhs_item_2'] + str(', ') + df_final['rhs_item_3']

df_final['lhs'] = df_final['lhs'].str.replace(',   ,  ', '').str.replace(',  ', ', ').str.strip()
df_final['rhs'] = df_final['rhs'].str.replace(',   ,  ', '').str.replace(',  ', ', ').str.strip()

df_final.head()
```

```python
df_final.columns = ['lhs_item_1','lhs_item_2','lhs_item_3','rhs_item_1','rhs_item_2','rhs_item_3','support','confidence','lift']
df_final.head()
```

```python
df_final.drop(columns=['lhs_item_1','lhs_item_2','lhs_item_3','rhs_item_1','rhs_item_2','rhs_item_3'],inplace=True)

df_final.head()
```
