**Training and Testing Data Split**

```python
y=df['Yearly Amount Spent']
X=df[['Avg. Session Length','Time on App','Time on Website','Length of Membership']]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split  (X,y,test_size=0.3,random_state=9108)
```

```python
y_train
```

```python
y_test
```

**Training the Model**

```python
from sklearn.linear_model import LinearRegression
lm=LinearRegression()
lm.fit(X_train,y_train)
```

```python
print('Coefficients:\n',lm.coef_)
```

```python
print('Intercept:\n',lm.intercept_)
```

**Predicting Test Data**

```python
predictions=lm.predict(X_test)
```

```python
predictions
```

**Evaluating The Model**

```python
from sklearn import metrics
print('MAE:',metrics.mean_absolute_error(y_test,predictions))
print('MSE:',metrics.mean_squared_error(y_test,predictions))
print('RMSE:',np.sqrt(metrics.mean_squared_error(y_test,predictions)))
```

**Residuals**

```python
import seaborn as sns
sns.distplot((y_test-predictions),bins=50)
plt.suptitle('Tanishka singh 9108',y=1.02)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Lasso
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import Ridge
```

```python
x=df[['Time on Website']]
y=df['Yearly Amount Spent']
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
degree=2
ploy_features=PolynomialFeatures(degree=degree,include_bias=False)
X_poly=ploy_features.fit_transform(X_train)
poly_reg=LinearRegression()
poly_reg.fit(X_poly,y_train)

X_test_poly=ploy_features.transform(X_test)
poly_pred=poly_reg.predict(X_test_poly)
```

```python
X_train
```

```python
X_poly
```

```python
#Linear Regression
linear_reg=LinearRegression()
linear_reg.fit(X_train,y_train)
linear_pred=linear_reg.predict(X_test)
```

```python
#Lasso Regression
alpha_lasso=0.1
lasso_reg=Lasso(alpha=alpha_lasso)
lasso_reg.fit(X_train,y_train)
lasso_pred=lasso_reg.predict(X_test)
```

```python
# Ridge Regression
ridge_reg = Ridge(alpha=1.0)  # You can change the regularization strength
ridge_reg.fit(X_train, y_train)
ridge_pred = ridge_reg.predict(X_test)

# Evaluate and compare the models
print("Linear Regression MSE:", mean_squared_error(y_test, linear_pred))
print("Polynomial Regression (Degree=2) MSE:", mean_squared_error(y_test, poly_pred))
print("Lasso Regression (alpha=1.0) MSE:", mean_squared_error(y_test, lasso_pred))
print("Ridge Regression (alpha=1.0) MSE:", mean_squared_error(y_test, ridge_pred))
```

```python
# Plot results
plt.scatter(X_test, y_test, color='black', label='Actual data')
plt.plot(X_test, linear_pred, color='red', label='Linear Regression')
plt.legend()
plt.xlabel('Time on website')
plt.ylabel('Yearly Amount Spent')
plt.title('Comparison of Regression Models')
plt.suptitle('Pooja ware 9118',y=1.02)
plt.show()
```
